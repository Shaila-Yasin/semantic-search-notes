# Import all Functions

```python
from utils.read_notes import load_documents
from utils.chunking import chunk_documents
from utils.similarity import semantic_search
from utils.embeddings import generate_embeddings
from utils.save_load import (
    save_embeddings,
    load_embeddings,
    embeddings_exist
)
```

# Save the Embeddings in a pickle file in `data` directory

```python
if embeddings_exist():

    print("Loading saved embeddings...")

    embedded_chunks = load_embeddings()

else:

    print("Generating embeddings...")

    documents = load_documents("notes")
    chunks = chunk_documents(documents)

    embedded_chunks = generate_embeddings(chunks)

    save_embeddings(embedded_chunks)
```

# Read the Notes 

```python
# Folder containing your notes
notes_folder = "notes"

# Load all documents
documents = load_documents(notes_folder)

print(f"Number of documents: {len(documents)}\n")

for doc in documents:
    print("=" * 50)
    print(f"File: {doc['filename']}")
    print("-" * 50)
    print(doc["content"][:200])   # Print first 200 characters
    print()
```

# Total chunks

```python
chunks = chunk_documents(documents)

print(f"Total Chunks: {len(chunks)}\n")

print(chunks[6])
```

# Total Embeddings

```python
embedded_chunks = generate_embeddings(chunks)

print(f"Total Embedded Chunks: {len(embedded_chunks)}")
```

## Read the material in embedding `30` and extract `10` first embeddings values

```python
chunk = embedded_chunks[30]

print("File:", chunk["filename"])
print("Title:", chunk["title"])
print("Text:", chunk["text"][:150])

print("\nEmbedding Length:", len(chunk["embedding"]))

print("\nFirst 10 Values:")
print(chunk["embedding"][:10])
```

# Ask a query and it will return top `3` most relevent chunks

```python
query = input("Ask a question: ")

results = semantic_search(query, embedded_chunks)
for result in results:

    print("=" * 60)
    print("Score :", round(result["score"], 4))
    print("File  :", result["filename"])
    print("Title :", result["title"])
    print("-" * 60)
    print(result["text"])
    print()
```
